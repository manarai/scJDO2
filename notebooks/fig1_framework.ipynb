Drawing only; no data. Content follows Results §1 of
`manuscript/scJDO_manuscript_v57_draft.md` (two operators, one operator
space, archetypes; identifiability components). Source task: none
(conceptual); the identifiability components link to the tasks that test
them (S12/S13 representation, S14 clock/k, S21b rotation, S10 stationarity
and diffusion, S18b composition).

```python
from _style import *
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

prov = Provenance("Fig1", notebook="notebooks/fig1_framework.ipynb")
prov.use("manuscript/scJDO_manuscript_v57_draft.md", panel="all", task="draft",
         note="text of Results section 1 (no numbers drawn)")
```

```python
fig, ax = plt.subplots(figsize=(DOUBLE_COL, 3.5))
fig.subplots_adjust(left=0.005, right=0.995, bottom=0.01, top=0.99)
ax.set_xlim(0, 100); ax.set_ylim(0, 60); ax.axis("off")

def box(x, y, w, h, text, fc="white", ec="#333333", fs=6.5, bold=False, tc="black"):
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.4,rounding_size=1.2",
                                fc=fc, ec=ec, lw=0.8))
    ax.text(x + w / 2, y + h / 2, text, ha="center", va="center", fontsize=fs,
            fontweight="bold" if bold else "normal", color=tc, linespacing=1.25)

def arrow(x0, y0, x1, y1, color="#333333", ls="-"):
    ax.add_patch(FancyArrowPatch((x0, y0), (x1, y1), arrowstyle="-|>", mutation_scale=8,
                                 lw=0.9, color=color, linestyle=ls))

# column titles
for x, t in [(10, "Time information"), (37, "Operator"), (64, "Operator space"), (89, "Archetypes")]:
    ax.text(x, 57.5, t, ha="center", fontsize=8, fontweight="bold")

# time-information sources
srcs = [("Snapshot", 47), ("Time course\n(sampling days)", 38.5), ("Lineage barcodes", 30),
        ("Metabolic labelling", 22.5), ("RNA velocity", 15)]
for t, y in srcs:
    box(2, y, 16, 5.5, t, fc="#F7F7F7")

# operators
box(25.5, 41, 23, 11, "Score operator\n$J_{score}=-\\beta\\nabla^2 E(z)$\nsymmetric; density curvature",
    fc="#FBE3D6", ec=C_SCORE, fs=6.2)
box(25.5, 13, 23, 17, "Flow operator\n$J_{flow}=\\partial f/\\partial z$\n$f$ fitted to a measured\nvelocity $v(z)$ (MeasuredDrift)\nlabel-free: entropic OT\nbetween days",
    fc="#D6E6F4", ec=C_FLOW, fs=6.2)
arrow(18.4, 49.7, 25.1, 47.5, C_SCORE)
for _, y in srcs[1:]:
    arrow(18.4, y + 2.7, 25.1, 21.5, C_FLOW)

# operator space
box(53.5, 20, 21, 26, "Basis-invariant descriptors\nper cell\n\nsign Re $\\lambda_1$, spectrum\nrotational index\nmax|Im $\\lambda$|/max|$\\lambda$|\nnon-normality (Henrici)\ntransient gain\n\nrank-transformed",
    fc="white", fs=6.2)
arrow(48.9, 46.5, 53.1, 38, C_SCORE)
arrow(48.9, 21.5, 53.1, 28, C_FLOW)
ax.text(64, 17.2, "score: $A^{-T}HA^{-1}$   flow: $AJA^{-1}$", ha="center", fontsize=6, color="#444444")

# archetypes
box(79.5, 26, 19.5, 14, "Recurrent patterns\n(semi-NMF on unit-norm J;\nK-means on descriptors)\nvs Leiden at matched K",
    fc="#EFEFEF", fs=6.2)
arrow(75.0, 33, 79.1, 33)

# identifiability strip
comps = [("Representation", "embedding metric\n(S12, S13)"),
         ("Clock", "time units, k\n(S14)"),
         ("Rotation", "antisymmetric flow\n(S21b, S15)"),
         ("Stationarity", "score = stability only\nat steady state (S10)"),
         ("Diffusion", "not identifiable at\n48-h sampling (S10)"),
         ("Composition", "proliferation/death,\noccupancy (S18b)")]
ax.text(50, 9.6, "Identifiability components (each tested in the Results)", ha="center",
        fontsize=7, fontweight="bold")
w = 15.4
for i, (h, d) in enumerate(comps):
    x = 2 + i * (w + 1.0)
    box(x, 0.6, w, 6.6, f"{h}\n{d}", fc="#FFF8E1", ec="#B08900", fs=6)

prov.panel("all", "Schematic: time information -> score/flow operator -> operator space -> archetypes; identifiability components", status="schematic")
save(fig, "Fig1")
```

```python
prov.legend("""
**Figure 1 | Two operators, one operator space.** Schematic (no data). Time information
determines which local operator can be estimated. A snapshot supports only the
**score operator** *J_score = −β∇²E(z)*, the Hessian of a neural energy fitted by score
matching; it is symmetric and describes density curvature. A **measured** velocity (entropic
optimal transport between sampling days, lineage displacement, metabolic labelling or RNA
velocity) supports the **flow operator** *J_flow = ∂f/∂z* of a smooth drift *f* fitted to that
velocity (MeasuredDrift). Per cell, basis-invariant descriptors (sign of Re λ₁, spectrum,
rotational index, non-normality, transient gain), standardised and rank-transformed, define an
operator space whose recurrent patterns are archetypes (compared with Leiden clustering at
matched K). Under a change of coordinates *A* the score operator transforms as *A⁻ᵀHA⁻¹* and
the flow operator as *AJA⁻¹*, so only descriptors invariant under both are compared directly.
Bottom: the six identifiability components and the tasks in which each is tested.
""")
prov.write()
```
